```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from xgboost import XGBClassifier
import pickle
```

```python
dataset = pd.read_csv('diabetes_prediction_dataset.csv.zip')
print(dataset.shape)
print(dataset['diabetes'].value_counts())
```

```python
# # Encode categorical variables
# le = LabelEncoder()
# dataset['gender'] = le.fit_transform(dataset['gender'])
# dataset['smoking_history'] = le.fit_transform(dataset['smoking_history'])
le_gender = LabelEncoder()
le_smoking = LabelEncoder()

dataset['gender'] = le_gender.fit_transform(dataset['gender'])
dataset['smoking_history'] = le_smoking.fit_transform(dataset['smoking_history'])
```

```python
# Split features and target
X = dataset.drop(columns='diabetes', axis=1)
Y = dataset['diabetes']
```

```python
dataset.isnull().sum()
```

```python
# Now, recreate X and Y after transforming 'gender'
X = dataset.drop(columns='diabetes', axis=1)
Y = dataset['diabetes']
print(X)
print(Y)
```

```python
# Train-test split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)

print(X_train.shape, X_test.shape)
```

```python
#imbalance ratio for scale_pos_weight
scale_pos_weight = len(Y_train[Y_train == 0]) / len(Y_train[Y_train == 1]) #no. positive /no. negative 
print(scale_pos_weight)
```

```python
#Model Training
classifier = XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,  # imbalance handling
    eval_metric='aucpr', #sensitive to minority class
    random_state=42
)
```

```python
classifier.fit(X_train, Y_train) #Trained here
```

```python
# Evaluate the Model
X_train_pred = classifier.predict(X_train)
X_test_pred = classifier.predict(X_test)

print("Training Accuracy:", accuracy_score(Y_train, X_train_pred))
print("Testing Accuracy:", accuracy_score(Y_test, X_test_pred))

print("\nClassification Report:\n", classification_report(Y_test, X_test_pred))
print("\nConfusion Matrix:\n", confusion_matrix(Y_test, X_test_pred))

# Format: (gender, age, hypertension, heart_disease, smoking_history, bmi, HbA1c_level, blood_glucose_level)
# Smoking:
# 0-No info
# 1-current
# 2-ever
# 3-former
# 4-never
# 5-not currently
# gender:
# 0-Female
# 1-Male
ip = (0, 58, 0, 0, 3, 46.05, 5, 90)
ip=np.asarray(ip)
ip_reshaped=ip.reshape(1,-1) #for only 1 data instance

prediction=classifier.predict(ip_reshaped)
print(prediction)

if(prediction[0]==0):
  print('Non Diabetic')
else:
  print('Diabetic')
```
